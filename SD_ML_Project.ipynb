## Stacking Classifier

```python
from sklearn.ensemble import StackingClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.ensemble import BaggingClassifier
from sklearn.ensemble import AdaBoostClassifier
```

## Hyperparameter Tuning
1. For Decision Tree what should be the max_depth
2. How many number of Decision Tree are required inside a Random Forest for best accuracy.

There are 2 ways one can perform Hyperparameter Tuning:
1. GridSearchCV
2. RandomSearchCV

## Grid Search

```python
from sklearn.model_selection import GridSearchCV
from sklearn.ensemble import RandomForestClassifier
para_grid={'n_estimators':[100,200],'max_depth':[4,6,8]}
grid=GridSearchCV(RandomForestClassifier(),para_grid,cv=5)
```

```python
grid.fit(X_train,Y_train)
```

```python
y_predict=grid.predict(X_test)
```

```python
print(grid.best_params_)
print(grid.best_score_)
```

## Random Search

```python
from sklearn.model_selection import RandomizedSearchCV
from sklearn.ensemble import RandomForestClassifier

para_grid={'n_estimators':[100,200,300],'max_depth':[4,6,8]}
random=RandomizedSearchCV(RandomForestClassifier(),param_distributions=para_grid,n_iter=10,cv=5)
```

```python
random.fit(X_train,Y_train)
```

```python
print(random.best_params_)
print(random.best_score_)
```

```python
from sklearn.ensemble import RandomForestClassifier
forest=RandomForestClassifier(n_estimators=200)
```

```python
forest.fit(X_train,Y_train)
```

```python
y_predict=forest.predict(X_test)
```

```python
accuracy_score(Y_test,y_predict)
```
